```python
import numpy as np
import pandas as pd
from sklearn.mixture import GaussianMixture
from sklearn.ensemble import RandomForestRegressor
from sklearn.multioutput import MultiOutputRegressor
import pickle
```

```python
input_df = pd.read_csv("data/combined.csv")
input_df.head()
```

```python
dates = set(input_df['acq_date'])
spatial_features = dict()
daily_rows = []
for d in dates:
    coords = input_df[input_df['acq_date']==d][['latitude', 'longitude']].values
    if len(coords)<10:
        continue
    gmm = GaussianMixture(n_components=10)
    gmm.fit(coords)

    row_data = {'date': d}
    weights = gmm.weights_
    for i, w in enumerate(weights):
        row_data[f'weight_{i}'] = w
    means = gmm.means_
    for i, m1 in enumerate(means):
        for j, m2 in enumerate(m1):
            row_data[f'mean_{i}_{j}'] = w
    covariances = gmm.means_
    for i, m1 in enumerate(means):
        for j, m2 in enumerate(m1):
            row_data[f'covariances_{i}_{j}'] = w
    daily_rows.append(row_data)
```

```python
df = pd.DataFrame(daily_rows)
```

```python
df['date'] = pd.to_datetime(df['date'])
df['day_of_year'] = df['date'].dt.dayofyear
df['month'] = df['date'].dt.month
df = df.drop(['date'], axis=1)
```
