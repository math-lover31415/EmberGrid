The above open meteo area is limited to limit data size. The irrelavant viirs data will be ignored

```python
# Since df data is rounded to nearest 0.25, we round viirs data as well
def round_quarter(x):
    return round(x*4)/4
viirs['longitude'] = viirs['longitude'].apply(round_quarter)
viirs['latitude'] = viirs['latitude'].apply(round_quarter)
```

```python
viirs['fire'] = 1
viirs.head()
```

```python
joined = df.merge(viirs, on=['latitude', 'longitude', 'date'], how='left')
joined.head()
```

```python
joined['fire'] = joined['fire'].fillna(0)
```
